```python

import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load IMDb dataset
(X_train_full, y_train_full), (X_test_full, y_test_full) = imdb.load_data(
    num_words=10000
)

# Combine training and testing data
X = list(X_train_full) + list(X_test_full)
y = np.concatenate([y_train_full, y_test_full])

# Select 2500 positive and 2500 negative reviews
positive_idx = np.where(y == 1)[0][:2500]
negative_idx = np.where(y == 0)[0][:2500]

selected_idx = np.concatenate([positive_idx, negative_idx])
np.random.seed(42)
np.random.shuffle(selected_idx)

X = [X[i] for i in selected_idx]
y = y[selected_idx]

print("Total reviews:", len(X))
print("Positive reviews:", np.sum(y == 1))
print("Negative reviews:", np.sum(y == 0))
```
